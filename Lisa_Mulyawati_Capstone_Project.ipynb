```python
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt

# Load dataset Titanic dari seaborn
df = sns.load_dataset("titanic")
df.head()
```

```python
# Cek struktur data
print(df.info())

# Cek statistik ringkas
print(df.describe(include="all"))
```

```python
# Jumlah penumpang selamat vs tidak selamat
sns.countplot(x="survived", data=df)
plt.title("Distribusi Survival Titanic (0=Tidak Selamat, 1=Selamat)")
plt.show()
```

```python
# Survival berdasarkan jenis kelamin
sns.countplot(x="sex", hue="survived", data=df)
plt.title("Survival Rate berdasarkan Gender")
plt.show()
```

```python
# Survival berdasarkan kelas
sns.countplot(x="class", hue="survived", data=df)
plt.title("Survival Rate berdasarkan Kelas Penumpang")
plt.show()
```

```python
# Hitung survival rate per gender
survival_gender = df.groupby("sex")["survived"].mean()
print("Survival Rate per Gender:\n", survival_gender)

# Hitung survival rate per class
survival_class = df.groupby("class")["survived"].mean()
print("\nSurvival Rate per Class:\n", survival_class)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt
import pandas as pd

# Load Titanic dataset dari seaborn
df = sns.load_dataset("titanic")

# Ambil hanya kolom numerik
numeric_df = df.select_dtypes(include=['float64', 'int64'])

# Hitung korelasi
corr = numeric_df.corr()

# Buat heatmap
plt.figure(figsize=(10,6))
sns.heatmap(corr, annot=True, cmap="coolwarm", center=0)
plt.title("Heatmap Korelasi Dataset Titanic", fontsize=14)
plt.show()
```
